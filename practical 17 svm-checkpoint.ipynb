```python
#%% chat gpt EXPERIMENT 17: LINEAR SVM
from sklearn.datasets import load_iris
from sklearn.svm import SVC
from sklearn.inspection import DecisionBoundaryDisplay
import matplotlib.pyplot as plt

iris = load_iris()
mask = iris.target != 2
X, y = iris.data[mask, :2], iris.target[mask]

svm = SVC(kernel="linear", C=1).fit(X, y)
plt.figure(figsize=(8, 6))
DecisionBoundaryDisplay.from_estimator(svm, X, response_method="decision_function",
                                       plot_method="contour", levels=[-1, 0, 1],
                                       colors="k", linestyles=["--", "-", "--"],
                                       ax=plt.gca())
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", edgecolors="k")
plt.scatter(svm.support_vectors_[:, 0], svm.support_vectors_[:, 1],
            s=150, facecolors="none", edgecolors="k", label="Support Vectors")
plt.title("Linear SVM Decision Boundary")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.svm import SVC

iris = load_iris()
X = iris.data[iris.target != 2, :2]
y = iris.target[iris.target != 2]

model = SVC(kernel='linear', C=1)
model.fit(X, y)

xx, yy = np.meshgrid(
    np.arange(X[:, 0].min()-0.5, X[:, 0].max()+0.5, 0.02),
    np.arange(X[:, 1].min()-0.5, X[:, 1].max()+0.5, 0.02)
)

Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
plt.scatter(model.support_vectors_[:, 0],
            model.support_vectors_[:, 1],
            s=150, facecolors='none', edgecolors='black')

plt.title('Linear SVM')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```
